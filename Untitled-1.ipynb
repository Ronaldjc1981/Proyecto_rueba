```python
%matplotlib inline
import matplotlib.pyplot as plt
import seaborn as sns

# Gráfica 1: Mapa de calor de correlaciones
plt.figure(figsize=(10,8))
corr_matrix = df[['HP','Attack','Defense','Sp. Atk','Sp. Def','Speed','Total']].corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0, fmt='.2f', cbar_kws={'label': 'Correlación'})
plt.title('Matriz de Correlación entre Estadísticas')
plt.tight_layout()
plt.show()

# Gráfica 2: Scatter plot Attack vs Defense coloreado por Type 1
plt.figure(figsize=(12,8))
top_6_types = df['Type 1'].value_counts().nlargest(6).index
scatter_df = df[df['Type 1'].isin(top_6_types)]
sns.scatterplot(data=scatter_df, x='Attack', y='Defense', hue='Type 1', palette='tab10', s=100, alpha=0.7)
plt.title('Attack vs Defense para los 6 Type 1 más comunes')
plt.xlabel('Attack')
plt.ylabel('Defense')
plt.legend(title='Type 1', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()

# Gráfica 3: Distribuciones múltiples de estadísticas
stats_to_plot = ['HP','Attack','Defense','Sp. Atk','Sp. Def','Speed']
fig, axes = plt.subplots(2,3, figsize=(15,10))
axes = axes.flatten()
for idx, stat in enumerate(stats_to_plot):
    axes[idx].hist(df[stat], bins=25, color='skyblue', edgecolor='black')
    axes[idx].set_title(f'Distribución de {stat}')
    axes[idx].set_xlabel(stat)
    axes[idx].set_ylabel('Frecuencia')
plt.tight_layout()
plt.show()

# Gráfica 4: Líneas de tendencia Speed vs Total por generación
plt.figure(figsize=(12,6))
for gen in sorted(df['Generation'].unique()):
    gen_data = df[df['Generation']==gen].sort_values('Speed')
    plt.plot(gen_data['Speed'], gen_data['Total'], marker='o', label=f'Gen {gen}', alpha=0.6)
plt.title('Relación Speed vs Total por Generación')
plt.xlabel('Speed')
plt.ylabel('Total')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# Gráfica 5: Radar chart de stats promedio por Legendary
from math import pi
fig, ax = plt.subplots(figsize=(10,10), subplot_kw=dict(projection='polar'))
stats = ['HP','Attack','Defense','Sp. Atk','Sp. Def','Speed']
angles = [n / float(len(stats)) * 2 * pi for n in range(len(stats))]
angles += angles[:1]
for legendary in [False, True]:
    values = df[df['Legendary']==legendary][stats].mean().tolist()
    values += values[:1]
    label = 'Legendario' if legendary else 'No Legendario'
    ax.plot(angles, values, 'o-', linewidth=2, label=label)
    ax.fill(angles, values, alpha=0.15)
ax.set_xticks(angles[:-1])
ax.set_xticklabels(stats)
ax.set_ylim(0, 100)
ax.set_title('Comparación de Stats Promedio: Legendary vs No Legendary', pad=20)
ax.legend(loc='upper right', bbox_to_anchor=(1.3, 1.1))
plt.tight_layout()
plt.show()
```
